# Automatic sentiment annotation script for the French-Fongbe dataset.

This script reads the cleaned French-Fongbe CSV file, uses a pre-trained
RoBERTa model optimized for French to predict the sentiment of the French column, and transfers the predicted labels to the Fongbe text.

```python
from pathlib import Path

import pandas as pd
from transformers import pipeline
from tqdm import tqdm
```

```python
# load dataset
csv_path = Path("../data/French_Fongbe_dataset3.csv")
df = pd.read_csv(csv_path)
df.head()
```

```python
classifier = pipeline("sentiment-analysis", model="cardiffnlp/twitter-xlm-roberta-base-sentiment", device=0)
batch_size = 64
texts = df["french"].tolist()
labels = []
```

```python
for i in tqdm(range(0, len(texts), batch_size)):
    batch = texts[i : i + batch_size]
    results = classifier(batch)
    labels.extend([res["label"] for res in results])
```

```python
df["labels"] = labels
```

```python
output_path = "../data/French_Fongbe_Sentiment_Dataset.csv"
df.to_csv(output_path, index=False, encoding="utf-8")
```
