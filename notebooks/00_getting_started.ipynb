# 00 — Getting Started with Tokeh

**Tokeh** is a concrete NLP learning project named after the *Calao* bird
(hornbill), known for its loud, distinctive call across West Africa.

This notebook walks through the core utilities provided by the package:
- Text preprocessing
- Vocabulary building
- TF-IDF vectorisation

```python
# Install the package if running outside the project root
# !pip install -e ..
```

## 1. Preprocessing

```python
from tokeh.preprocessing import clean_text, simple_tokenize, remove_stopwords

raw = "The quick brown fox jumps over the lazy dog."
cleaned = clean_text(raw)
print('Cleaned :', cleaned)

tokens = simple_tokenize(cleaned)
print('Tokens  :', tokens)

tokens_no_sw = remove_stopwords(tokens)
print('No stops:', tokens_no_sw)
```

## 2. Vocabulary & Embeddings

```python
from tokeh.embeddings import build_vocab, one_hot_encode, tfidf_vectorize

corpus = [
    simple_tokenize(clean_text("NLP is a fascinating field")),
    simple_tokenize(clean_text("Deep learning powers modern NLP")),
    tokens_no_sw,
]

vocab = build_vocab(corpus, min_freq=1)
print(f'Vocabulary size: {len(vocab)}')
print('Sample entries:', dict(list(vocab.items())[:6]))
```

```python
vec = one_hot_encode('nlp', vocab)
print('One-hot for "nlp":', vec)
```

```python
import pandas as pd

matrix = tfidf_vectorize(corpus, vocab)
df = pd.DataFrame(matrix, columns=list(vocab.keys()))
df
```
